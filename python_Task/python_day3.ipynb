```python
#1Q
s = "python is easy and python is powerful"
words = s.split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
```

```python
#2Q
word = "banana"
frequency = {}

for ch in word:
    frequency[ch] = frequency.get(ch, 0) + 1

print(frequency)
```

```python
#3Q
word = "swiss"
result = None

for ch in word:
    if word.count(ch) == 1:
        result = ch
        break

print(result)
```

```python
#4Q
s1 = "listen"
s2 = "silent"

if sorted(s1) == sorted(s2):
    print(True)
else:
    print(False)
```

```python
#5Q
s = "programming"
seen = set()
result = None

for ch in s:
    if ch in seen:
        result = ch
        break
    seen.add(ch)

print(result)
```

```python
#6Q
nums = [0, 3, 0, 1, 2, 0, 4]

result = []

for num in nums:
    if num != 0:
        result.append(num)

zeros = nums.count(0)
result.extend([0] * zeros)

print(result)
```

```python
#7Q
nums = [10, 20, 40, 30, 40, 20]

unique = list(set(nums))
unique.sort()

if len(unique) >= 2:
    print(unique[-2])
else:
    print(None)
```

```python
#8Q
nums = [4, 2, 4, 5, 2, 7, 5]
result = []

for num in nums:
    if num not in result:
        result.append(num)

print(result)
```

```python
#9Q
nums = [1, 2, 4, 5, 6]
N = 6

expected_sum = N * (N + 1) // 2
actual_sum = sum(nums)

missing = expected_sum - actual_sum

print(missing)
```

```python
#10Q
nums = [2, 7, 11, 15]
target = 9
result = []

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            result = [nums[i], nums[j]]
            break
    if result:
        break

print(result)
```
