```python
#8Q
s = "banana".replace(" ", "")
frequency = {}

for ch in s:
    frequency[ch] = frequency.get(ch, 0) + 1

most_frequent = s[0]

for ch in s:
    if frequency[ch] > frequency[most_frequent]:
        most_frequent = ch

print(most_frequent)
```
